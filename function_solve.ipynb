```python
def gcd (a,b):
  if b==0:
    return a
  else:
    return gcd(b,a%b)
def lcm (a,b):
  return (a*b)//gcd (a,b)
a= int(input("Enter the firs value: "))
b= int(input("Enter the second value: "))
print("GCD=",gcd (a,b))
print("LCM=",lcm(a,b))
```

```python
#Answer:27
def multi_check(num1, num2):
    if num1 % 3 == 0 and num2 % 3 == 0:
        return "Both numbers are divisible by 3"
    elif num1 % 2 == 0 or num2 % 2 == 0:
        return "At least one number is even"
    else:
        return "Neither condition is satisfied"
num1 = int(input("Enter your number1: "))
num2 = int(input("Enter your number2: "))
print(multi_check(num1, num2))
```
